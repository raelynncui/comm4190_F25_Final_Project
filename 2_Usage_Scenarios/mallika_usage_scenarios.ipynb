# Usage Scenarios: Mallika

# Scenario 1: Personalized Reading Lesson for a Student with Dyslexia

## Short Description:
Creating a customized reading comprehension plan for a student with dyslexia.

## Narrative:
A teacher uses AdaptEd to generate a lesson plan for Maya, a 4th-grader who struggles with decoding complex words. The app asks the teacher for details about Maya’s reading level, preferred learning style, and goals. The LLM then produces a phonics-based reading activity, simplified text, and progress check questions.

## Step-by-Step Interaction:
1. Teacher opens AdaptEd and selects “Reading Support.”  
2. Inputs: “Maya, 4th grade, dyslexia, focus on reading fluency.”  
3. LLM generates: lesson plan, activity sheet, comprehension quiz.  
4. Teacher reviews, edits, and saves lesson.  
5. AdaptEd tracks completion and performance feedback.  

## Data Description:
- Input: Student profile (age, disability type, goal, reading level).  
- Output: Generated lesson content, progress metrics, teacher edits.
  
## Evaluation:
Success is measured by whether the lesson matches difficulty, student engagement improves, and teacher accepts ≥80% of LLM’s content.

Possible issues: over-simplification, biased assumptions about learning disability, or teacher input ambiguity.

# Scenario 2: Parent Tracking Weekly Math Progress for Child with ADHD

## Short Description
Helping a parent monitor and adapt math lessons for a child with ADHD.

## Narrative
A parent logs weekly progress for Sam, who has trouble focusing during math practice. AdaptEd suggests short, gamified tasks and visual aids, then adjusts next week’s plan based on completion data.

## Step-by-Step Interaction:
1. Parent inputs Sam’s attention span and last week’s results.  
2. LLM recommends bite-sized, timed math exercises.  
3. Parent logs completion data.  
4. AdaptEd updates the plan and provides motivational feedback.
   
## Data Description:
- Input: Behavior logs, task completion rates, engagement notes.  
- Output: Updated learning plan, engagement trend visualization.
- 
## Evaluation:
Success would be if adaptive pacing improves consistency.

Possible errors: incomplete logs, overfitting to short-term fluctuations, or misinterpretation of engagement data.
