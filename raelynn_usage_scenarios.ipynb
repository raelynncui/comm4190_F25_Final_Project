# Usage Scenarios: Raelynn

## Scenario 1: Reading plan for a dyslexic 4th grader

### Description
Creating a standards-aligned reading lesson plan that adapts weekly to a student’s dyslexia profile and assessments.

### Narrative of scenario
A 4th-grade teacher needs to design a reading plan for a 4th grader who has dyslexia. The teacher is short on time and overwhelmed by juggling different grade-level standards, accommodations, and monitoring the progress of other students. In AdaptEd, the teacher enters the student’s profile (strengths, decoding challenges, Lexile band, goals, time constraints). AdaptEd then proposes a plan with decodable texts, multi-sensory activities, read-aloud supports, and weekly quick checks. After each week, the system will update its difficulty, flags any potential regressions, and explains why adjustments are recommended.

### Step by step breakdown
1. Teacher opens the student’s profile and selects “Create weekly plan.”
2. Inputs: reading level, target standards, IEP (Individualized Education Program) objectives, available minutes/day, accommodations, and any materials on hand.
3. AdaptEd generates:
- Weekly plan
- Printable decodable passages and comprehension questions (at the student’s level)
- Multi-sensory activities
- Progress assessments ("probes")
- Rationale + alternative options
4. Teacher tweaks (to keep a human in the loop) and assigns.
5. At the end of week: enters probe scores or uploads audio. AdaptEd updates plan and shows recommendations.

### Data

Data needed:
- Student profile: grade, reading level, IEP (Individualized Education Program) goals, accommodations, strengths/challenges, constraints
- Curriculum standards: state standards by grade
- Prior performance: probe scores, notes, audio samples
- Context: class schedule, materials available, device/printing constraints

Data created:
- Lesson plan: goals, activities, timings, and notes
- Materials: generated passages, and question sets
- Assessments: different probes and scoring rubrics
- Analytics: weekly trends, mastery predictions, and rationale/explanations

### Evaluation

Success criteria:
- Plan quality: the rubric should be aligned to IEP goals/standards and provide an appropriate level of text difficulty and accommodations. The quality should meet/exceed some defined threshold by expert raters.
- Time saved: compare the median time to an effective plan compared to the teacher baseline (what they were doing before)
- Student outcomes: there should be an improvement in weekly probe scores.

Complications and handling errors:
- Sparse data: this application of AdaptEd requires that enough data regarding the student is provided to tailor the lesson plan effectively.
- Accessibility: materials may not be screen-reader friendly. There should be a dyslexia-friendly font option.

## Scenario 2: Parent working with a 7th grader with ADHD

### Description
Guided at-home mini-lessons with timers, breaks, and behavior supports that adapt nightly based on engagement and completion signals.

### Narrative of scenario
A parent struggles with their 7th grade child completing homework due to distractinos, procrastination, andmeltdowns. They need a simple, consistent routine with supports that actually fit their child’s attention needs. In AdaptEd, the parent sets a 45-minute evening window, selects today’s subjects, and notes triggers (such as noise). AdaptEd then proposes a Pomodoro-style plan (for example, a 10-min math review, then a 2-min break, then a 12-min worksheet, then a 3-min movement break, and finally 8-min reading, with visual timers, step-by-step prompts, and praise scripts). If the child abandons a step, the app will suggest a shorter period of working time or another form of learning (oral response instead of written, for example). After the session, the parent logs friction points and any difficulties they faced. After, the next day’s plan auto-adjusts based on this feedback.

### Step-by-step breakdown

1. Parent opens AdaptEd and picks subjects, sets window, selects attention supports.
2. AdaptEd proposes a time-boxed flow with micro-goals and visuals.
3. During work blocks, the app shows on-screen timers, chunked instructions, and a text-to-speech option.
4. The app offers breaks (movement suggestions + quick checks like “Ready to resume?”).
5. If stall detected (no interaction 60s or parent taps “struggling”), it recommends de-escalation tips and offers reduced chunks or alternate modalities (oral instead of written, for example).
6. End: a quick reflection (emoji scale), completion log, note triggers, and a small reward suggestion.
7. The next session auto-adjusts durations, ordering, and supports based on this reflection.

### Data 

Needed:
- Student attention profile, such as their typical focus span and triggers
- Educational context such as what subjects the student is learning and upcoming deadlines
- Parent preferences for how they want to structure the learning session

Created:
- A learning routine plan with time-stampled blocks, durations, instructions, break types, and prompts
- An engagement log based on what the student completed
- Adaptation parameters based on the parent's reflection at the end of the session

### Evaluation

Success criteria:
- Routine completion rate increased
- If parents qualitatively report that they feel less of a burden and quantitatively rport decrease time spend negotiating/redirecting with their child
- Increased positive student sentiments throughout the sessions

Complications and handling errors:
- Parents and students may experience sensory overload with these frequent timers. This could be mitigated with a low-stimulus UI and configurable prompt frequency.
- Conflicts with school accommodations. This could be mitigated with an “IEP aware mode” that highlights potential conflicts and asks the parent to confirm.
- The possibility of unexpected inputs (“do 4 hours nonstop”) requires guardrails to cap block length and ensure healthy learning patterns.
